## Agregamos filas y columnas

### Para agregar filas y columnas a una matriz de numpy podemos usar la funcion append y se usa de la siguiente manera.

### np.append(< Matriz >,< Nuevo valor >,< tipo de insercion >)

### < tipo de insercion >: Si este valor es 0 se agregan los nuevos valores como una fila y si es un 1 se agregan como columnas.

```python
np.append(M,[[7,8,9]], 0)
```

```python
np.append(M,[[7],[8]], 1)
```

### Tambien podemos usar la funcion insert la cual nos va a permitir agregar valores en un indice de la matriz

### np.insert(< Matriz >, <índice>, < Nuevo valor > , < tipo de insercion >)

### < tipo de insercion >: Si este valor es 0 se agregan los nuevos valores como una fila y si es un 1 se agregan como columna

```python
A
```

```python
np.insert(A, 0, [100, 200, 300], 0)
```

```python
np.insert(A, 0, [100, 200, 300], 1)
```

## Recorriendo una matriz numpy

### Podemos usar el for para recorrer una matriz numpy.

### Ocupamos dos "for", uno para movernos en las filas y otro para movernos en las columnas.


```python
M
```

### Veamos un ejemplo en el que queremos sumar los valores de cada fila.

```python
M
```

```python
tam_filas, tam_columnas = M.shape
sumas_por_fila = []
suma = 0

for indice_fila in range(tam_filas):
    for indice_columna in range(tam_columnas):
        suma = suma + M[indice_fila, indice_columna]
    sumas_por_fila.append(suma)
    suma = 0

sumas_por_fila
```
