## Practica #7. Geometría de la imagen, Rotación, escala y traslación

### Las transformaciones geométricas de una imagen son técnicas de procesamiento digital que permiten modificar el tamaño, la orientación y la posición de los objetos en la imagen espacialmente. En esta practica diseñaremos un programa en python que implemente operaciones de rotación, escalado y traslación utilizando la librería PIL.

### Paso 1. Añadir las librerias necesarias.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

### Paso 2. Funciones para aplicar transformaciones de Rotación, Escala y Traslación.

```python
def rotar_imagen(imagen, angulo):
    # Usamos la funcion rotate de PIL. expand=True adapta el tamaño para no cortar la imagen.
    return imagen.rotate(angulo, expand=True)

def escalar_imagen(imagen, factor_escala):
    # Calculamos el nuevo tamaño
    ancho, alto = imagen.size
    nuevo_ancho = int(ancho * factor_escala)
    nuevo_alto = int(alto * factor_escala)
    # Aplicamos resize
    return imagen.resize((nuevo_ancho, nuevo_alto))

def trasladar_imagen(imagen, tx, ty):
    # Utilizamos transform con transformacion AFIN (AFFINE)
    # La matriz para traslación es (1, 0, -tx, 0, 1, -ty)
    matriz = (1, 0, -tx, 0, 1, -ty)
    return imagen.transform(imagen.size, Image.AFFINE, matriz)
```

### Paso 3. Cargar la Imagen y aplicar las transformaciones

```python
# Cargamos la imagen original usando tu ruta preferida
imagen_original = Image.open("C:/Users/silva/Downloads/3.jpg")

# Mostramos la imagen original
plt.imshow(imagen_original)
plt.title("Imagen Original")
plt.axis('off')
plt.show()
```

### Paso 3.1. Aplicar Rotación

```python
angulo = 45 # Rotar 45 grados
imagen_rotada = rotar_imagen(imagen_original, angulo)

plt.imshow(imagen_rotada)
plt.title(f"Rotación de {angulo}°")
plt.axis('off')
plt.show()
```

### Paso 3.2. Aplicar Escala

```python
factor = 1.5 # Escalar al 150%
imagen_escalada = escalar_imagen(imagen_original, factor)

plt.imshow(imagen_escalada)
plt.title(f"Escala factor {factor}")
plt.axis('off')
plt.show()
```

### Paso 3.3. Aplicar Traslación

```python
desplazamiento_x = 100
desplazamiento_y = 50
imagen_trasladada = trasladar_imagen(imagen_original, desplazamiento_x, desplazamiento_y)

plt.imshow(imagen_trasladada)
plt.title(f"Traslación X={desplazamiento_x}, Y={desplazamiento_y}")
plt.axis('off')
plt.show()
```
