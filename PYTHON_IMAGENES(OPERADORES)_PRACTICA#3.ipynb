## Imagen Digital.

## Difgitalmente, una imagen puede representarse como una matriz, especialmente cuando hablamos de procesamiento de imagenes.

### Procesamiento Digital de Imagenes.

### El procesamiento digital de imagenes es el conjunto de tecnicas que se aplican a las imagenes digitales con el obtejtivo de mejorar la calidad o facilitar la busqueda de información.

## Practica No. 3. Operadores individuales (Negativo, Escala de Grices, potencia (Oscurecer, Abrillantar una imagen)) y la binarización de imagenes digitales.

### En esta practica diseñaremos un programa de Python para implementar los operadores individuales en el procesamiento digital de imagenes.

## Paso 1: Importar las librerias necesarias

```python
from PIL import Image
import matplotlib.pyplot as plt
```

## Paso 2: Cargar la imagen

```python
imagen = Image.open("C:\\Users\\maluc\\Downloads\\EddySavage-GoldenMonkeysBlog-4.jpg")
```

## Paso 3: Obtener las dimensiones de la imagen

```python
ancho,alto = imagen.size
ancho, alto
```

## Paso 4: Crear una nueva imagen para guardar el negativo.

```python
negativa = Image.new("RGB", (ancho, alto))
```

## Paso 5: Recorrer la matriz de pixeles

```python
for x in range(ancho):
    for y in range(alto):

        # Obtener los valores RGB del pixel
        r, g, b = imagen.getpixel((x, y))

        # Calcular el negativo
        r_nuevo = 255 -r
        g_nuevo = 255 -g
        b_nuevo = 255 -b

        # Guardar el nuevo pixel
        negativa.putpixel(
            (x,y),
            (r_nuevo, g_nuevo, b_nuevo)

        )
```

## Paso 6: Guardar la imagen negativa

```python
negativa.save("imagen_negativa.jpg")
imagen_negativa = Image.open("imagen_negativa.jpg")
```

## Paso 7: Mostrar la imagen original y la imagen negativa utilizando matplotlib

```python
plt.figure(figsize = (10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.title("Imagen Original")
plt.axis('off') # Ocultar ejes

# Mostrar la imagen negativa
plt.subplot(1, 2, 2)
plt.imshow(imagen_negativa)
plt.title("Imagen Negativa")
plt.axis('off') # Ocultar ejes

# Mostrar ambas imagenes
plt.show()
```

## Paso 8: Recorrer la matriz de pixeles(Imagen en gris)

```python
for x in range(ancho):
    for y in range(alto):
        r, g, b = imagen.getpixel((x, y))

        # Calcular escala de grises
        gris = int((r + g + b) / 3)

        # Guardar el nuevo píxel
        negativa.putpixel((x, y), (gris, gris, gris))
```

## Paso 9: Guardar la imagen gris

```python
negativa.save("imagen_gris.jpg")
imagen_gris = Image.open("imagen_gris.jpg")
```

## Paso 10: Mostrar la imagen original y la imagen gris utilizando matplotlib

```python
plt.figure(figsize=(10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.title("Imagen Original")
plt.axis('off')

# Mostrar la imagen procesada
plt.subplot(1, 2, 2)
plt.imshow(imagen_gris)
plt.title("Imagen en Grises")
plt.axis('off')

plt.show()
```

## Paso 11: Recorrer la matriz de pixeles(Imagen potenciada en abrillantamiento)

```python
for x in range(ancho):
    for y in range(alto):
        r, g, b = imagen.getpixel((x, y))

        r = int(255 * ((r/255) ** 0.25))
        g = int(255 * ((g/255) ** 0.25))
        b = int(255 * ((b/255) ** 0.25))

        # Guardar el nuevo píxel
        negativa.putpixel((x, y), (r, g, b))
```

## Paso 12: Guardar la imagen potenciada(Abrillantada)

```python
negativa.save("imagen_potenciada.jpg")
imagen_potenciada = Image.open("imagen_potenciada.jpg")
```

## Paso 13: Mostrar la imagen original y la imagen potenciada utilizando matplotlib

```python
plt.figure(figsize=(10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.title("Imagen Original")
plt.axis('off')

# Mostrar la imagen procesada
plt.subplot(1, 2, 2)
plt.imshow(negativa)
plt.title("Imagen Potenciada")
plt.axis('off')

plt.show()
```

## Paso 14: Recorrer la matriz de pixeles(Imagen potenciada en oscurecimiento)

```python
for x in range(ancho):
    for y in range(alto):
        r, g, b = imagen.getpixel((x, y))

        r = int(255 * ((r/255) ** 2))
        g = int(255 * ((g/255) ** 2))
        b = int(255 * ((b/255) ** 2))

        # Guardar el nuevo píxel
        negativa.putpixel((x, y), (r, g, b))
```

## Paso 15: Guardar la imagen potenciada(Oscurecida)

```python
negativa.save("imagen_potenciada.jpg")
imagen_potenciada = Image.open("imagen_potenciada.jpg")
```

## Paso 16: Mostrar la imagen original y la imagen potenciada utilizando matplotlib

```python
plt.figure(figsize=(10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.title("Imagen Original")
plt.axis('off')

# Mostrar la imagen procesada
plt.subplot(1, 2, 2)
plt.imshow(negativa)
plt.title("Imagen Potenciada")
plt.axis('off')

plt.show()
```

## Paso 17: Recorremos matriz de pixeles para la binarización

```python
umbral = 128
binaria = Image.new("RGB", (ancho, alto))

for x in range(ancho):
    for y in range(alto):
        r, g, b = imagen.getpixel((x, y))

        gris = int((r + g + b) / 3)

        if gris >= umbral:
            nuevo_valor = 255
        else:
            nuevo_valor = 0

        binaria.putpixel((x, y), (nuevo_valor, nuevo_valor, nuevo_valor))
```

## Paso 18 Mostramos la imagen original y la binarizada

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen)
plt.title("Imagen Original")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(binaria)
plt.title(f"Imagen Binarizada, Umbral: {umbral}")
plt.axis("off")

plt.show()
```
