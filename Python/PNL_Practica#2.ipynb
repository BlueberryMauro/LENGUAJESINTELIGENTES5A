## ¿Qué es el Procesamiento del lennguaje natural (PLN o NLP en ingles)?

### El procesamiento de Lenguaje Natural(PLN) es una rama de la Inteligencia Artificial que permite que las computadoras entiendad, interpreten y generen lenguaje humano (ya sea hablado o escrito)

### En otras palabras, el PLN busca que una computadora pueda leer, escuchar o escribir texto de una forma que tenga sentido para los humanos

## Ejemplos comunes del PLN:

### 1.Asistentes virtuales como Alexa, Siri o ChatGPT.
### 2.Traductores automaticos (como Google Translate)
### 3.Analisis de sentimientos(detectar si un texto es positivo o negativo).
### 4.Motores de busqueda
### 5. Chatbots de atencion al cliente, etc...

## Practia #2. En esta practica implementaremos un segundo ejemplo de PNL (por sus siglas en el idioma Ingles), utilizando la libreria nltk

### word_tokenize(). La funcion word_tokenize() de python sirve para dividir un texto en palabras y signos de puntuacion, para que podamos analizarlo más facilmente en tareas de Procesamiento del Lenguaje Natural(PLN)

### stopwords(). Las stopwords son palabras muy comunes en un idioma que normalmente no aportan significado importante al analisis del texto, por ejemplo:

### En español: el, la, los, las, de, que, y, a, en...

### En ingles: the, is, at, on, for, and...

## Paso 1: Importar las librerias necesarias

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

## Paso 2. Definir un texto de ejemplo.

```python
texto = "La inteligencia artificial está transformando el mundo de la tecnologia"
```

## Paso 3. Dividir el texto en palabras(tokens)

```python
tokens = word_tokenize(texto)
print("Tokens:", tokens)
```

## Paso 4: Eliminar las stopwords (palabras vacias como: 'el', 'de', 'que'...)

```python
stop_words = set(stopwords.words('spanish'))
tokens_filtrados = [palabra for palabra in tokens if palabra.lower() not in stop_words]
print("Tokens sin stopword:", tokens_filtrados)
```
