### El WordCloud (nube de palabras) sirve para visualizar de forma grafica las palabras más frecuentes en un texto.
### El WordCloud toma un texto, cuenta cuantas veces aparece cada palabra, y luego muestra esas palabras en una imagen donde:

### Las palabras más grandes son las que más se repiten
### Las palabras pequeñas son las menos frecuentes

## En esta practica implementaremos un ejemplo de Procesamiento de Lenguaje Natural utilizando la libreria NTKL para contar cuantas veces aparece cada palabra y la graficamos utilizando WordClaud

## Paso 1: Importar las librerias necesarias

```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from wordcloud import WordCloud
from collections import Counter
```

## Paso 2. Definir un texto de ejemplo

```python
text = """
La inteligencia artificial es una tecnologia que permite a las computadoras aprender de los datos.
El machine learning utiliza algoritmos para aprender patrones y realizar predicciones.
La inteligencia artificial se utiliza en medicina, educacion, industria, robotica, y procesamiento de datos.
El machine learning permite desarrollar sistemas intelligentes que pueden aprender y mejorar.
"""
```

## Paso 3. Dividir el texto en palabras (tokens)

```python
# Convertir el texto a minusculas
texto = text.lower()

# Seperar el texto en palabras utilizando NTLK
palabras = word_tokenize(texto,language = "spanish")

print("\nPalabras despues de la limpieza: ")
print("palabras")
```

## Paso 4. Eliminar las stopwords(palabras vacias como: 'el', 'de', 'que',...)

```python
# Obtener las palabras que no apartan informacion

stop_words = set(stopwords.words('spanish'))


# Eliminar signos de puntuacion y stopwords

palabras_limpias = [
    palabra for palabra in palabras
    if palabra.isalpha() and palabra not in stop_words
]

print("\nPallabras despues de la limpieza: ")
print(palabras_limpias)
```

## Paso 5. Contar la frecuencia de las palabras

```python
frecuencias = Counter(palabras_limpias)

print("\nFrecuencia despues de la limpieza: ")
print(frecuencias)
```

## Paso 6: Generar la nube de palabras

```python
nube = WordCloud(
    width=1000,
    height=800,
    background_color="white",
    colormap = "viridis",
).generate_from_frequencies(frecuencias)
```

## Paso 7: Mostrar la nube de palabras

```python
plt.figure(figsize=(12,8))
plt.imshow(nube, interpolation="bilinear")
plt.axis("off")
plt.title("Nube de palabras: Inteligencia Artificial")
plt.show()
```
