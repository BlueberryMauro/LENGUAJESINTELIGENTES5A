### En esta practica implementaremos un programa en Python para proteger un aplicación LoT usando autentitación básoca y detección de accesos sospechoosos. La idea es evitar que cualquier dispositivo no autorizado se conecte.

## Paso 1: Importar las librerias necesarias

```python
import os
import pandas as pd
import numpy as np
```

## Paso 2: Seguridad Basica para IOT (Dispsitivos Autorizados)

```python
dispositivos_autorizados = {
    "sensor_01": "abc123",
    "sensor_02": "xyz783"
}
```

## Paso 3: Controlar el numero de intentos

```python
intentos_fallidos = {}
```

## Paso 4: Definir una Funcion de Autenticacion

```python
def autenticar_dispositivo(id_dispositivo, token):
    if id_dispositivo in dispositivos_autorizados:

        if dispositivos_autorizados[id_dispositivo] == token:
            print(f"✅ Acceso permitido a {id_dispositivo}")
            return True

    # Acceso sospechoso
    registrar_intento(id_dispositivo)
    print(f"❌ Acceso denegado a {id_dispositivo}")
    return False
```

## Paso 5: Definar una Funcion para Registrar los intentos sospechosos

```python
def registrar_intento(id_dispositivo):
    if id_dispositivo not in intentos_fallidos:
        intentos_fallidos[id_dispositivo] = 0
    intentos_fallidos[id_dispositivo] += 1
    
    # Bloque simple
    if intentos_fallidos[id_dispositivo] >= 3:
        print(f"🚨 Dispositivo {id_dispositivo} BLOQUEADO por intentos sospechosos")
```

## Paso 6: Implementamos la Conexion IOT

```python
autenticar_dispositivo("sensor_01", "abc123")  # Acceso permitido
autenticar_dispositivo("sensor_01", "wrong_token")  # Acceso denegado
autenticar_dispositivo("sensor_03", "123456")  # no resgistrado
autenticar_dispositivo("sensor_03", "123456")  # intento repetido
autenticar_dispositivo("sensor_03", "123456")  # bloqueado
```
