### El WordCloud (nube de palabras) sirve para visualizar de forma gráfica las palabras más frecuentes en un texto.
El WordCloud toma un texto, cuenta cuántas veces aparece cada



## Paso 1: Importar las librerias necesarias.

```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from prometheus_client.utils import INF
from wordcloud import WordCloud
from collections import Counter
import os
```

## Paso 2: Definir un texto de ejemplo:

```python
carpeta = "/home/adolfo/PycharmProjects/LenguajesInteligentes/palabrasComunes/"

textos = []

for archivo in ["12AgryMen.txt", "darkKnight.txt", "fightClub.txt", "goodBadUgly.txt", "goodfather.txt", "goodfather2.txt", "redemption.txt", "pulpFiction.txt"]:
    ruta = os.path.join(carpeta, archivo)
    with open(ruta, "r", encoding="latin-1") as f:
        textos.append(f.read())

texto = "\n".join(textos)
```

## Paso 3. Dividir el texto en palabras (tokens).

```python
# Convertir el texto a minúsculas
texto = texto.lower()

# Separar el texto en palabras utilizando NLTK

palabras = word_tokenize(texto)
```

## Paso 4. Eliminar las stopwords

```python
# Obtener las palabras que no aportan información
stop_words = set(stopwords.words('english'))

# Eliminar signos de puntuación y stopwords
palabras_limpias = [
    palabra for palabra in palabras
    if palabra.isalpha() and palabra not in stop_words
]
```

## Paso 5. Contar la frecuencia de las palabras.

```python
frecuencias = Counter(palabras_limpias)
```

## Paso 6. Generar la nube de palabras.

```python
import numpy as np
from wordcloud import WordCloud

nube = WordCloud(
    width=800,
    height=800,
    background_color="white",
    colormap="viridis",
    max_words=None
).generate_from_frequencies(frecuencias)
```

## Paso 7. Mostrar la nube de palabras.

```python
plt.figure(figsize=(12,12))
plt.imshow(nube, interpolation='bilinear')
plt.axis("off")
plt.title("Nube de palabras: ")
plt.show()
```
