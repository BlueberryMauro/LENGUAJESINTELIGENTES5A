## Practica #5. Filtro de Colores

### Los filtros de colores de una imagen son técnicas de procesamiento digital que permiten modificar, seleccionar o resaltar determinados colores de una imagen para mejorar su apariencia o facilitar su análisis. En esta practica diseñamos un programa en python que implemente filtros de colores.

### Paso 1. Añadir las librerias necesarias.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

### Paso 2. Función para aplicar un filtro de colores a los canales R, G y B.

```python
def aplicar_filtro_rgb(imagen, filtro_r, filtro_g, filtro_b):
    # 1. Convertir explícitamente a modo RGB (3 canales: R, G, B)
    imagen_rgb = imagen.convert('RGB')

    # 2. Convertir a array de NumPy usando float32 para evitar desbordamiento al multiplicar
    arr = np.array(imagen_rgb, dtype=np.float32)

    # 3. Aplicar el filtro multiplicando cada canal
    arr[:, :, 0] = np.clip(arr[:, :, 0] * filtro_r, 0, 255)  # Canal R
    arr[:, :, 1] = np.clip(arr[:, :, 1] * filtro_g, 0, 255)  # Canal G
    arr[:, :, 2] = np.clip(arr[:, :, 2] * filtro_b, 0, 255)  # Canal B

    # 4. Convertir de vuelta a uint8 y regresar la imagen filtrada
    return Image.fromarray(arr.astype(np.uint8))
```

### Paso 3. Cargar la Imagen y aplicar el filtro de color con diferentes valores para R, G y B

### Paso 3.1. Filtro de Color Rojo

```python
# Cargar la imagen
imagen_original = Image.open("C:/Users/maluc/Downloads/mapache.png")

# Aplicar el filtro de color
filtro_r = 1.5  # Aumentar el canal rojo
filtro_g = 0
filtro_b = 0

imagen_filtrada = aplicar_filtro_rgb(imagen_original, filtro_r, filtro_g, filtro_b)
```

```python
# Mostrar la imagen original y la imagen con el filtro aplicado utilizando matplotlib
plt.figure(figsize=(10, 5))

# Mostrar la imagen original
plt.subplot(1, 2, 1)
plt.imshow(imagen_original)
plt.title('Imagen Original')
plt.axis('off') # Ocultar los ejes

# Mostrar la imagen con el filtro aplicado
plt.subplot(1, 2, 2)
plt.imshow(imagen_filtrada)
plt.title(f"Filtro RGB (R={filtro_r}, G={filtro_g}, B={filtro_b})")
plt.axis('off') # Ocultar ejes

# Mostrar ambas imagenes
plt.show()
```
